# Simple Linear Regression - Mathematical Formulation

## 1. The Goal
In Simple Linear Regression, we are given a dataset with one input column (e.g., `CGPA`) and one output column (e.g., `Package`). The data is "sort of linear," so our goal is to find the **Best Fit Line** that minimizes the errors across all data points.

The equation of this line is:
$$y = mx + b$$
Our entire mathematical goal is to find the optimal values for **$m$** (slope/weight) and **$b$** (y-intercept/offset).

## 2. Two Ways to Solve for 'm' and 'b'
1. **Closed-Form Solution (OLS - Ordinary Least Squares):** This is a direct mathematical formula. It uses a finite sequence of operations (addition, subtraction, multiplication, etc.) to arrive at the exact values of $m$ and $b$. Scikit-Learn's `LinearRegression()` class uses this behind the scenes. It works beautifully for smaller dimensions (like our 2D data).
2. **Non-Closed-Form Solution (Gradient Descent):** An approximation algorithm that uses calculus (differentiation) to step iteratively toward the optimal $m$ and $b$. It is heavily used in high-dimensional spaces or Deep Learning where the OLS formula becomes computationally too expensive. Scikit-Learn's `SGDRegressor()` class uses this method.

Today, we are formulating the **OLS (Closed-Form) Solution** from scratch.

## 3. Formulating the Error Function
To find the "best" line, we must first mathematically define what a "bad" line is. We do this by calculating the **Error** (or Loss Function).

* **The Error per point:** For any single student, the error is the difference between their *Actual Package* ($y_i$) and the *Predicted Package* ($\hat{y}_i$).
    $$Error_i = y_i - \hat{y}_i$$
* **Squaring the Error:** We square the error $(y_i - \hat{y}_i)^2$. Why? 
    1. It prevents negative errors (points below the line) from canceling out positive errors (points above the line).
    2. Squaring heavily penalizes large outliers.
    3. Unlike the Absolute Value function $|x|$, a square function is a smooth, continuous parabola, making it easily differentiable using calculus.

* **Total Error Function (Loss Function):** The sum of all squared errors for $n$ data points.
    $$E = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

Since we know that $\hat{y}_i = m(x_i) + b$, we substitute it into the Error equation:
$$E(m, b) = \sum_{i=1}^{n} \left[ y_i - (m x_i + b) \right]^2$$

Notice that the error $E$ is entirely a function of $m$ and $b$. The $x_i$ and $y_i$ values are static data points.

## 4. Minimizing the Error (Calculus)
If you plot the Error $E$ against $m$ and $b$ on a 3D graph, it forms a massive bowl shape (a convex function). Our goal is to reach the absolute bottom of this bowl, where the error is minimum. 

From calculus, we know that the minimum point of a function occurs where the slope (derivative) is exactly zero. Because we have two variables ($m$ and $b$), we must calculate the **Partial Derivatives** and set them to zero:
1. $\frac{\partial E}{\partial b} = 0$
2. $\frac{\partial E}{\partial m} = 0$

### Solving for 'b'
Using the chain rule on $\frac{\partial E}{\partial b} = 0$:
$$b = \bar{y} - m\bar{x}$$
Where $\bar{y}$ is the mean of all actual outputs, and $\bar{x}$ is the mean of all inputs.

### Solving for 'm'
Using the chain rule on $\frac{\partial E}{\partial m} = 0$, and substituting the derived value of $b$ into the equation, we get the final formula for the slope:
$$m = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^{n} (x_i - \bar{x})^2}$$

These are the exact mathematical formulas that dictate how a Simple Linear Regression model learns!
