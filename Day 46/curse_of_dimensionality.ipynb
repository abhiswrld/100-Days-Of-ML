# 1. What is curse of dimensionality?

<p> It can be said that this is basically curse of features, basically the more we add features, it doesn't necessarily mean that the performance will increase, and can also decrease.

<p> There is something called an optimal amount of features.

While adding a few relevant features initially helps the model learn better, continuously adding features eventually introduces noise, causing the model to perform worse.

## 2. Why is High-Dimensional Space a Problem?
* **Sparsity of Data:** As you add more dimensions, the volume of the feature space grows exponentially. If you don't also exponentially increase your amount of training data, the data points become scattered and isolated (sparse).
* **Distance becomes Meaningless:** In high-dimensional spaces, the distance between any two data points becomes roughly the same. This completely breaks distance-based machine learning algorithms (like KNN or K-Means) because they can no longer distinguish between "near" and "far" neighbors.

## 3. The Impact on Machine Learning Models
1. **Performance Decrease:** The model starts capturing noise and random correlations in the high-dimensional space instead of true patterns, leading to severe overfitting.
2. **Computational Complexity:** More columns mean more calculations. The time and memory required to train the model increase drastically, sometimes making it impossible to train.

## 4. How to Solve It? (Dimensionality Reduction)
To fight the curse, we use Dimensionality Reduction techniques. There are two main approaches:

### A. Feature Selection
Selecting a smaller, optimal subset of the original features and dropping the rest. 
* **Techniques:** Forward Selection, Backward Elimination.
* **Note:** This is usually done *after* studying all machine learning algorithms, as you need a trained model to evaluate which features are actually useful.

### B. Feature Extraction
Creating completely new, independent features that capture the core essence (variance) of all the original features combined, allowing us to compress a 100-column dataset into just a few new columns.
* **Techniques:** PCA (Principal Component Analysis), LDA (Linear Discriminant Analysis), t-SNE.
