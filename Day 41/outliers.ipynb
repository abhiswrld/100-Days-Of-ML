# Day 41: Outliers in Machine Learning

## 1. What are Outliers?
Outliers are data points that differ significantly from other observations. They lie far away from the central tendency of the data. While most data points follow a specific trend or cluster together, outliers are isolated points far removed from the general pattern.

## 2. When is an Outlier Dangerous?
An outlier is not always bad; it depends on the context.
*   **Data Errors:** Sometimes outliers are simply mistakes in data entry (e.g., entering an Age of 300) These must be handled or removed.
*   **Anomaly Detection:** In fields like fraud detection, medical diagnosis, or network security, the outlier itself is the event of interest (an anomaly). In these cases, outliers are crucial to the model.

## 3. Effect of Outliers on ML Algorithms
Not all machine learning algorithms are affected by outliers equally.
*   **Sensitive Algorithms (Weight-based):** Algorithms that rely on updating weights or calculating distances are generally very sensitive to outliers. Examples include:
    *   Linear Regression
    *   Logistic Regression
    *   AdaBoost
    *   Deep Learning / Neural Networks
*   **Robust Algorithms (Tree-based):** Algorithms that make splits based on thresholds (like creating branches) are largely unaffected by outliers. Examples include:
    *   Decision Trees
    *   Random Forest
    *   Gradient Boosting Trees

## 4. How to Detect Outliers
The detection technique depends largely on the distribution of your data:

### A. Normally Distributed Data (Z-Score Treatment)
If the feature follows a normal (Gaussian) distribution, we use the Empirical Rule. Data points that fall outside of 3 standard deviations from the mean are considered outliers.
*   **Lower Limit:** $\mu - 3\sigma$ (Mean - 3 * Standard Deviation)
*   **Upper Limit:** $\mu + 3\sigma$ (Mean + 3 * Standard Deviation)

### B. Skewed Data (IQR Based Filtering)
For skewed distributions, the Z-score method is less effective. Instead, we use the Interquartile Range (IQR).
*   **IQR** = Q3 (75th percentile) - Q1 (25th percentile)
*   **Lower Limit:** $Q1 - 1.5 \times IQR$
*   **Upper Limit:** $Q3 + 1.5 \times IQR$

### C. Percentile Based Approach
This method is distribution-agnostic and relies on setting a percentile threshold. For instance, you might consider any value below the 1st percentile or above the 99th percentile as an outlier.

## 5. How to Treat Outliers
Once detected, outliers can be treated using several techniques:

1.  **Trimming:** Simply dropping the rows that contain outliers. This is fast but can lead to a "thin" dataset if too many rows are removed, resulting in a loss of valuable data.
2.  **Capping (Winsorization):** Capping the outliers at a specific threshold (e.g., setting all values above the 99th percentile to the exact value of the 99th percentile). This retains the data points but limits their extreme influence on the model.
3.  **Treating as Missing Values:** Replacing the extreme outlier values with NaN and then using standard imputation techniques (like mean/median imputation or KNN Imputer) to fill them.
4.  **Discretization:** Binning continuous variables into discrete groups (e.g., putting all ages from 90-100 into a single category). This naturally neutralizes the effect of extreme continuous values.
