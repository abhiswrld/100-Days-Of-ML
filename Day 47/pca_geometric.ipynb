# Day 47: Principal Component Analysis (PCA) - Geometric Intuition

## 1. What is PCA?
Principal Component Analysis (PCA) is an **unsupervised machine learning algorithm** and the most famous **Feature Extraction** technique. Its primary goal is to solve the Curse of Dimensionality by transforming high-dimensional data into lower-dimensional data while retaining the "essence" (variance) of the original data.

### Two Main Benefits of PCA:
1.  **Faster Execution:** By reducing the number of features (e.g., from 10 columns to 3 columns), the dataset becomes smaller, allowing machine learning algorithms to train and predict much faster while saving computational power.
2.  **Data Visualization:** Humans can only visualize up to 3 dimensions. PCA allows us to compress high-dimensional data (like 100 features) into 2D or 3D, making it possible to plot and understand the data visually.

## 2. Feature Selection vs. Feature Extraction
To understand PCA, we must understand why Feature Selection fails in certain scenarios:
*   **Feature Selection:** You analyze the original features and keep only the ones with the highest variance (spread). For example, if "Number of Rooms" has a high variance but "Number of Grocery Shops" has almost no variance, you keep Rooms and drop Grocery Shops.
*   **The Problem:** What if you have two features (e.g., "Number of Rooms" and "Number of Washrooms") that have an equal amount of variance and a linear relationship? You cannot drop either without losing important information.

**This is where Feature Extraction (PCA) comes in.** Instead of dropping one of the columns, PCA creates a *completely new* column (e.g., "Size of Flat") that combines the information of both.

## 3. The Geometric Intuition
If you have a 2D scatter plot where the data is spread equally across both the X and Y axes, PCA will mathematically rotate the coordinate system.

1.  **Rotation:** PCA rotates the axes so that the data points align along a new set of axes.
2.  **Principal Components:** These new axes are called **Principal Components** (PC1, PC2, etc.). You will always get the same number of Principal Components as your original features (e.g., 2 original columns = 2 Principal Components).
3.  **Maximizing Variance:** PCA specifically aligns the first Principal Component (PC1) along the path where the data has the *maximum spread (variance)*.
4.  **Extraction:** Once the axes are rotated, PC1 captures the majority of the data's variance. You can then safely drop PC2, effectively reducing the data from 2D to 1D without losing the relationship between the data points.

## 4. Why is Variance So Important?
Variance is proportional to the **spread** of the data. When projecting data from a higher dimension to a lower dimension, if you do not maximize the variance, the data points will overlap or squish too closely together.
*   Distance-based algorithms (like KNN) rely on the distance between data points to make classifications.
*   By maximizing the variance on the new Principal Components, PCA ensures that the data points remain separated in the lower dimension, preserving their true relationships and distances.
